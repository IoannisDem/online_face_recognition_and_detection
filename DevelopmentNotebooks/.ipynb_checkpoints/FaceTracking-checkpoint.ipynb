```python
from facenet_pytorch import MTCNN, InceptionResnetV1, fixed_image_standardization, training
import torch
import cv2
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plt
from skimage.metrics import structural_similarity as ssim
import time
```

## Load model and tracker

```python
# set device
device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
print('Current device: {}'.format(device))
```

```python
mtcnn = MTCNN(
    image_size=160,
    margin=0,
    min_face_size=20,
    thresholds=[0.6, 0.7, 0.7],
    factor=0.709,
    post_process=True,
    device=device
)
```

## Functions

## Start recording

```python
cap = cv2.VideoCapture(0)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print(width, height)
```

```python
multiTracker = cv2.legacy.MultiTracker_create()
tracker = cv2.legacy.TrackerKCF_create()
```

```python
threshold = 0.9
frame_counter = 0

prev = np.zeros([height, width]).astype('uint8')
# prev = cv2.cvtColor(prev, cv2.COLOR_BGR2GRAY) # make it a black picture
ssim_thresh = 0.85

ssims = []

start = time.time()
while True:
    ret, frame = cap.read()
    
    # Update tracker
    success, bboxes = multiTracker.update(frame)
    
    current = frame.copy()
    current = cv2.cvtColor(current, cv2.COLOR_BGR2GRAY)
    
    if not frame_counter%5:
        ssims.append(ssim(current, prev))
        prev = current.copy()
        switcher = True
    
        # major changes in image
        if ssims[-1] < ssim_thresh and switcher:
            print('start')

            # detect faces within the frame
            faces, probs = mtcnn.detect(frame[:,:,::-1])

            # run only if faces are detected
            if probs.any():
                print('Found')
                thres_probs = np.where(probs>threshold)
                bboxes = []
                for idx in thres_probs[0]:
                    print(idx)
                    x0, y0, x1, y1 = faces[idx].astype(int)
                    w = x1 - x0
                    h = y1 - y0

                    bbox = (x0, y0, w, h)
                    bboxes.append(bbox)

                    #initiate tracker
#                     ok = tracker.init(frame, (x0, y0, w, h))
                    multiTracker.add(tracker, frame, bbox)
                    success = True
                    # plot rectangle
    #                 cv2.rectangle(frame, (x0, y0), (x1, y1), (255, 0, 0,), 3)
        switcher=False
        
    if success:
        # Tracking success
        for bbox in bboxes:
            p1 = (int(bbox[0]), int(bbox[1]))
            p2 = (int(bbox[0] + bbox[2]), int(bbox[1] + bbox[3]))
            cv2.rectangle(frame, p1, p2, (255,0,0), 2, 1)
    
    cv2.imshow('Tracked person', frame)
    
    # Exit if ESC pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    frame_counter += 1
    
    
    
end = time.time()
cv2.waitKey(0)
cv2.destroyAllWindows()
```
