## Start recording

```python
ksize = (5, 5)
frame_count = 0
red    = (0, 0, 255)
yellow = (0, 255, 255)
green  = (0, 255, 0)

cap = cv2.VideoCapture(0)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print(width, height)
```

```python
def convert_errode(frame):
    frame_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    thresh = cv2.adaptiveThreshold(frame_gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 21, 10)
    frame_erode = cv2.erode(thresh, np.ones(ksize, np.uint8))
    return frame_erode
```

```python
while True:
    ret, frame = cap.read()
    
    if not frame_count:
        prev_frame = convert_errode(frame)
        frame_count += 1
    
    else:
        frame_erode = convert_errode(frame)
        error = np.mean(np.sqrt(frame_erode/255 - prev_frame/255))
        
        cv2.imshow('Contours', frame_erode)
        print(error)
        # Exit if ESC pressed
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
    prev_frame = frame_erode
    
cv2.waitKey(0)
cv2.destroyAllWindows()
```
